## Sample descriptives

```python
sample_FOI_descriptives(
    data=FOI_data,
    table_mode="ON",
    indicator_family="NUMBER_LAS",
    type="mixed",
    indicator=None,
    html="ON",
)
```

* `table_mode`: `"ON"` prints every indicator in the chosen family and type, and `indicator` must be `None`. `"OFF"` prints the chosen indicator.
* `indicator_family`: `"NUMBER_LAS"` (V1–V8) or `"TOTAL_EXPENDITURE"` (V9–V23).
* `type`: for `NUMBER_LAS`, `"mixed"`, `"count"`, or `"prevalence"`. For `TOTAL_EXPENDITURE`, `"mixed"`, `"gross_exp"`, or `"pct_exp"`.
* `indicator`: one code when `table_mode="OFF"`. For expenditure `"mixed"`, a pair of one code from V9–V17 and one from V18–V23.
* `html`: `"ON"` opens the same table in the browser so it can be copied. `"OFF"` leaves it in the notebook.

`NUMBER_LAS`. `"mixed"` prints each count with its share of the unique local authorities. `"count"` prints the counts. `"prevalence"` prints the shares.

```python
sample_FOI_descriptives(data=FOI_data)
sample_FOI_descriptives(data=FOI_data, type="count")
sample_FOI_descriptives(data=FOI_data, type="prevalence")
sample_FOI_descriptives(data=FOI_data, table_mode="OFF", indicator="V8")
```

`TOTAL_EXPENDITURE`. `"gross_exp"` prints V9–V17. `"pct_exp"` prints V18–V23. `"mixed"` prints both.

```python
sample_FOI_descriptives(

sample_FOI_descriptives(

sample_FOI_descriptives(

    indicator_family="TOTAL_EXPENDITURE",

    indicator=("V12", "V18"),
)
```

**NUMBER_LAS**

* **V1** count: Number of unique LAs in the sample
* **V1** prevalence: Prevalence of unique LAs in the sample
* **V2** count: Number of LAs that answered Q1 with a total expenditure figure, including zero
* **V2** prevalence: Prevalence of unique LAs answering Q1 with a total expenditure figure, including zero
* **V3** count: Number of LAs that answered Q2 with a total expenditure figure, including zero
* **V3** prevalence: Prevalence of unique LAs that answered Q2 with a total expenditure figure, including zero
* **V4** count: Number of LAs that answered both Q1 and Q2 with total expenditure figures, including zero
* **V4** prevalence: Prevalence of unique LAs that answered both Q1 and Q2 with total expenditure figures, including zero
* **V5** count: Number of LAs that answered N/A for Q1
* **V5** prevalence: Prevalence of unique LAs that answered N/A for Q1
* **V6** count: Number of LAs that answered N/A for Q2
* **V6** prevalence: Prevalence of unique LAs that answered N/A for Q2
* **V7** count: Number of LAs that answered N/A for both Q1 and Q2
* **V7** prevalence: Prevalence of unique LAs that answered N/A for both Q1 and Q2
* **V8** count: Number of LAs that answered zero expenditure (whole nil return) for Q1
* **V8** prevalence: Prevalence of unique LAs that answered zero expenditure (whole nil return) for Q1

**TOTAL_EXPENDITURE**

* **V9** gross_exp: Total gross expenditure for Q1
* **V10** gross_exp: Total gross expenditure for Q2
* **V11** gross_exp: Total gross expenditure for Q1 and Q2
* **V12** gross_exp: Total gross expenditure in scope for Q1
* **V13** gross_exp: Total gross expenditure in scope for Q2
* **V14** gross_exp: Total gross expenditure in scope for Q1 and Q2
* **V15** gross_exp: Total gross expenditure out of scope for Q1
* **V16** gross_exp: Total gross expenditure out of scope for Q2
* **V17** gross_exp: Total gross expenditure out of scope for Q1 and Q2
* **V18** pct_exp: Percentage of total Q1 gross expenditure that is in scope (V12 / V9)
* **V19** pct_exp: Percentage of total Q2 gross expenditure that is in scope (V13 / V10)
* **V20** pct_exp: Percentage of total gross expenditure that is in scope (V14 / V11)
* **V21** pct_exp: Percentage of total Q1 gross expenditure that is out of scope (V15 / V9)
* **V22** pct_exp: Percentage of total Q2 gross expenditure that is out of scope (V16 / V10)
* **V23** pct_exp: Percentage of total gross expenditure that is out of scope (V17 / V11)
