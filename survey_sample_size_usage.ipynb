# Survey Sample Size Calculation

This notebook shows how to use `calculate_survey_sample_size()` from `functions.py` for a cross-sectional survey estimating category prevalence.

## Assumptions

- Study design: percentage/prevalence outcome in a cross-sectional survey
- Confidence level: 95%
- Prevalence: 50%, the conservative assumption
- Population size: 24,453 schools in England
- Cluster sampling: no
- Margin of error: 5%
- Anticipated response rate: 50%

```python
from functions import calculate_survey_sample_size
```

```python
sample_size = calculate_survey_sample_size(
    anticipated_response_rate=0.3
)
sample_size
```

The two headline results are:

- `completed_responses_required`: the number of completed survey responses needed
- `schools_to_invite`: the number of schools to invite after adjusting for the anticipated response rate

```python
print(f"Completed responses required: {sample_size['completed_responses_required']}")
print(f"Schools to invite: {sample_size['schools_to_invite']}")
```

## Changing assumptions

You can pass different values if a specific category has a different expected prevalence, if you want a narrower margin of error, or if the response rate assumption changes.

```python
alternative_scenario = calculate_survey_sample_size(
    prevalence=0.20,
    margin_of_error=0.05,
    anticipated_response_rate=0.40,
)

alternative_scenario
```

## Cluster sampling example

The current survey assumption is no cluster sampling. If cluster sampling is introduced, set `cluster_sampling=True` and provide a `design_effect`.

```python
clustered_scenario = calculate_survey_sample_size(
    cluster_sampling=True,
    design_effect=1.5,
)

clustered_scenario
```
